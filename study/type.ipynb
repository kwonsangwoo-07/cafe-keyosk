```python
print("식: True")
print("타입:", type(True))
print("값:", True)

print("식: False")
print("타입:", type(False))
print("값:", False)

print("식: 2 > 3")
print("타입:", type(2 > 3))
print("값:", 2 > 3)

print("식: 2 < 3")
print("타입:", type(2 < 3))
print("값:", 2 < 3)

print("식: 3 == 3")
print("타입:", type(3 == 3))
print("값:", 3 == 3)
```
